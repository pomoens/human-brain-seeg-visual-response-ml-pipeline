# Event exporter

This script takes as input the events (stored in .txt file exported from brainstorm) with their onset in samples and category criteria. It outputs a txt file containing all event based on selected criteria.

 0 is onset in samples and column 1 is event code

```python
import numpy as np
```

```python
events_path = r"C:\ ... \Events.txt"

events = []
with open(events_path, 'r') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        onset, code = line.split('\t')
        events.append((int(onset), int(code)))

events = np.array(events)

print(f"Loaded {len(events)} events")
```

```python
def get_events_by_code(*code_inputs):

    all_matching = []

    for code_input in code_inputs:
        type_code    = int(code_input/100)
        method_code  = int((code_input%100)/10)
        emotion_code = int(code_input%10)
    
        #search in three stages, once for every code:
        matching_events = np.copy(events)
        if type_code != 0 :
            #eliminate all codes from matching_events that do not match with this type
            matching_events = matching_events[(matching_events[:, 1] // 100) == type_code]
    
        if method_code !=0 :
            #eliminate all codes from matching_events that do not match with this method
            matching_events = matching_events[((matching_events[:, 1] % 100) // 10) == method_code]
    
        if emotion_code != 0 :
            #eliminate all codes from mathcing events that do not match with this emotion
            matching_events = matching_events[(matching_events[:, 1] % 10) == emotion_code]
    
        print(f"Found {len(matching_events)} events matching input {code_input}")
        all_matching.append(matching_events)

    #turn list of arrays (all_matching) into single array (combined) and sort chronologically by sample
    combined = np.concatenate(all_matching, axis=0)
    combined = combined[combined[:, 0].argsort()]
    
    print(f"Total: {len(combined)} events")
    return combined
```

Example for all onset of type faces and static method

```python
events_arr = get_events_by_code(310)
```
